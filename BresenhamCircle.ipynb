```python
import matplotlib.pyplot as plt
```

```python
def bresenham_circle(r):
    x = 0
    y = r
    p = 3 - 2 * r

    points = []
    table = []
    i = 0

    while x <= y:

        old_x = x
        old_y = y
        old_p = p

        points.append((x, y))
        points.append((-x, y))
        points.append((x, -y))
        points.append((-x, -y))
        points.append((y, x))
        points.append((-y, x))
        points.append((y, -x))
        points.append((-y, -x))

        if p < 0:
            case = "Case I"

            new_p = p + 4 * x + 6
            x = x + 1

        else:
            case = "Case II"

            new_p = p + 4 * (x - y) + 10
            x = x + 1
            y = y - 1

        table.append([
            i,
            old_x,
            old_y,
            old_p,
            case
        ])

        p = new_p
        i += 1

    return points, table
```

```python
r = int(input("Enter radius: "))

points, table = bresenham_circle(r)

print("\nBresenham's Circle Drawing Algorithm")
print("Radius =", r)

print("\n")
print("{:<5} {:<5} {:<5} {:<8} {:<10}".format(
    "i", "x", "y", "P", "Case"
))

print("-" * 40)

for row in table:
    print("{:<5} {:<5} {:<5} {:<8} {:<10}".format(
        row[0], row[1], row[2], row[3], row[4]
    ))

x_points = []
y_points = []

for x, y in points:
    x_points.append(x)
    y_points.append(y)

plt.scatter(x_points, y_points)

plt.title("Bresenham's Circle Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid(True)
plt.axis("equal")

plt.show()
```
